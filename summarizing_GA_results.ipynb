```python
import pandas as pd 

dataset_target = 'bank_notes'

wd = f'XRF_databases/{dataset_target}/plsda/smeX_ga_optimization_hof.csv'
GA_results = pd.read_csv(wd, sep=';')

# vamos extrair o resumo estatistico de acordo com cada seed (coluna Seed)
# Fitness deve ser media, agregador deve ser moda, metrica deve ser moda, n bags deve ser media, frac_sample_per_bag deve ser media, frac_min_samples_per_predicate deve ser media e replacement deve ser moda
summary_df = GA_results.groupby('Seed').agg({
    'Fitness_RBO_Score': 'mean',
    'Agregador': lambda x: x.mode()[0],
    'Métrica': lambda x: x.mode()[0],
    'N_Bags': 'mean',
    'Frac_Samples_per_Bag': 'mean',
    'Frac_Min_Samples_per_Predicate': 'mean',
    'Replacement': lambda x: x.mode()[0]
}).reset_index()
summary_df
```
