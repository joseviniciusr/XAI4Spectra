```python
import pandas as pd 

dataset_target = 'bank_notes'

wd = f'XRF_databases/{dataset_target}/plsda/smeX_ga_optimization_hof.csv'
GA_results_bank_notes = pd.read_csv(wd, sep=';').groupby('Seed').head(10).reset_index(drop=True)
GA_results_bank_notes
```

```python

# vamos extrair o resumo estatistico de acordo com cada seed (coluna Seed)
# Fitness deve ser media, agregador deve ser moda, metrica deve ser moda, n bags deve ser media, frac_sample_per_bag deve ser media, frac_min_samples_per_predicate deve ser media e replacement deve ser moda
summary_bank_notes = GA_results_bank_notes.groupby('Seed').agg({
    'Fitness_RBO_Score': 'mean',
    'Agregador': lambda x: x.mode()[0],
    'Metric': lambda x: x.mode()[0],
    'N_Bags': 'mean',
    'Frac_Samples_per_Bag': 'mean',
    'Frac_Min_Samples_per_Predicate': 'mean',
    'Replacement': lambda x: x.mode()[0]
}).reset_index()
summary_bank_notes
```

## Gráficos comparativos por parâmetro, seed e dataset

```python
# Imports para visualização
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style='whitegrid', context='notebook')

# Juntar os resultados brutos (cada seed tem várias linhas – melhor para distribuições)
df_all_raw = pd.concat([
    GA_results_bank_notes.assign(Dataset='bank_notes'),
    GA_results_soil.assign(Dataset='soil'),
    GA_results_ecigar.assign(Dataset='ecigar'),
    GA_results_soil_types.assign(Dataset='soil_types')
], ignore_index=True)

# Colunas numéricas a analisar
numeric_params = ['Fitness_RBO_Score', 'N_Bags', 'Frac_Samples_per_Bag', 'Frac_Min_Samples_per_Predicate']

# --- PLOTS NUMÉRICOS EM DUAS COLUNAS ---
import numpy as np
num_numeric = len(numeric_params)
cols = 2
rows = int(np.ceil(num_numeric / cols))
fig, axes = plt.subplots(rows, cols, figsize=(13, 2*rows))
axes = axes.flatten() if num_numeric > 1 else [axes]

for idx, param in enumerate(numeric_params):
    pivot = df_all_raw.groupby(['Seed', 'Dataset'])[param].mean().unstack(level=-1)
    ax = axes[idx]
    sns.heatmap(pivot, annot=True, fmt='.3f', cmap='vlag', center=pivot.values.mean(), ax=ax)
    ax.set_title(f'Média de {param} por Seed x Dataset')
    ax.set_ylabel('Seed')
    ax.set_xlabel('Dataset')

# Remove subplots não usados
for j in range(idx+1, len(axes)):
    fig.delaxes(axes[j])
plt.tight_layout()
plt.show()

# --- PLOTS CATEGÓRICOS EM DUAS COLUNAS ---
cat_params = ['Agregador', 'Metric', 'Replacement']
num_cat = len(cat_params)
cols = 2
rows = int(np.ceil(num_cat / cols))
fig, axes = plt.subplots(rows, cols, figsize=(13, 2*rows))
axes = axes.flatten() if num_cat > 1 else [axes]

for idx, param in enumerate(cat_params):
    # pivot com a moda (categoria mais frequente) por Seed x Dataset
    mode_pivot = df_all_raw.groupby(['Seed', 'Dataset'])[param].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else 'NA').unstack(level=-1)
    # Contagem de vezes que a modal categoria apareceu em cada grupo Seed x Dataset
    counts_pivot = df_all_raw.groupby(['Seed', 'Dataset'])[param].agg(lambda x: int(x.value_counts().iloc[0]) if not x.value_counts().empty else 0).unstack(level=-1)
    # preparar anotações: "categoria (n=8)"
    annot = mode_pivot.fillna('NA').astype(str) + ' (n=' + counts_pivot.fillna(0).astype(int).astype(str) + ')'
    # mapear categorias para inteiros para colorização, mantendo rótulos nas anotações
    unique_vals = sorted(pd.unique(mode_pivot.values.ravel()))
    unique_vals = [u for u in unique_vals if (u is not None) and (str(u) != 'nan') and (u != 'NA')]
    cat2int = {c: i for i, c in enumerate(unique_vals)}
    mode_pivot_num = mode_pivot.replace(cat2int)
    ax = axes[idx]
    sns.heatmap(mode_pivot_num, annot=annot, fmt='', cmap='Pastel1', cbar=False, ax=ax)
    ax.set_title(f'Moda de {param} por Seed x Dataset (categoria + contagem)')
    ax.set_ylabel('Seed')
    ax.set_xlabel('Dataset')
for j in range(idx+1, len(axes)):
    fig.delaxes(axes[j])
plt.tight_layout()
plt.show()
```
