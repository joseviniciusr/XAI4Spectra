# SMeX (Spectral Model eXplanator)
* **INPUT**: $X$ spectral dataset which the model was trained upon, $y$ model predictions, $Z=Z[z_1, ...,z_M]$ spectral zones (provided by expert knowledge or automatically), $A$ aggregation method, $Q=[q_1, ...,q_K]$ quantiles, $S$ number of samples to randomly select for bagging, $N$ number of bags to create.

* **for** $m=1$ to $M$ **do**<br>
Extract $z_m$ spectral zone from $X$ and aggregate its values employing method $A$ to create aggregated zone values $z_m^{agg}$ 
    * **for** $k=1$ to $K$ **do**<br>
    Calculate $q_k$ quantile of $z_m^{agg}$, create predicate $P_{m,k}$ covering $z_m^{agg}$ and $y$ upon associated samples and store in a $P=[P_{m,1},...P_{m,K},...,P_{M,K}]$ list of predicates<br>
* Initialize directed graph $G = (V, E)$ with terminal nodes $\{Class_A, Class_B\}$<br>
* **for** $n=1$ to $N$ **do**<br>
Randomly select $S$ samples from $X$, filter predicates in $P$ based on $S$ to create bag $b_n$, and store bags in a $B=[b_1,...,b_N]$ list of bags<br>
Calculate association metric $\rho(P_{i}, y)$ between each predicate $P_i$ in bag $b_n$ and model predictions $y$, rank predicates by $\rho$ values in descending order to create ranking $R_n=[P_{i_1}, P_{i_2}, ..., P_{i_j}]$ where $\rho(P_{i_1}) \geq \rho(P_{i_2}) \geq ... \geq \rho(P_{i_j})$<br>
Add sequential edges $(P_{i_k}, P_{i_{k+1}})$ for consecutive predicates in ranking $R_n$, assign edge weight $w(P_{i_k}, P_{i_{k+1}}) = \rho(P_{i_k})$ and accumulate weights for repeated edges<br>
* Calculate Local Reaching Centrality $LRC(P_i)$ for each predicate node $P_i$ in graph $G$<br>
* **OUTPUT**: Ranked list $L=[P_{j_1}, P_{j_2}, ..., P_{j_l}]$ where $LRC(P_{j_1}) \geq LRC(P_{j_2}) \geq ... \geq LRC(P_{j_l})$ representing global importance of spectral zone predicates
